```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D

# Data
data1 = {
    'Models': ['LinearRegression', 'Random Forest', 'XGBoost', 'MLPRegressor', 'SVR', 'CatBoostRegressor', 'LGBMRegressor'],
    'MAE': [4.906, 2.739, 2.826, 5.234, 5.061, 2.454, 2.76],
    'Runtime [ms]': ['0:00:43.387145', '0:28:11.761681', '0:03:58.883474', '0:01:44.252276',
                     '0:04:52.754769', '0:19:36.925169', '0:04:51.223103']
}
data2 = {
    'Models': ['LinearRegression', 'Random Forest', 'XGBoost', 'MLPRegressor', 'SVR', 'CatBoostRegressor', 'LGBMRegressor'],
    'MAE': [4.575, 2.345, 2.129, 4.414, 4.353, 2.281, 2.511],
    'Runtime [ms]': ['0:00:45.055854', '0:10:55.468473', '0:01:01.575033', '0:00:31.231719',
                     '0:02:12.258870', '0:08:16.526615', '0:15:25.084937']
}
data3 = {
    'Models': ['LinearRegression', 'Random Forest', 'XGBoost', 'MLPRegressor', 'SVR', 'CatBoostRegressor', 'LGBMRegressor'],
    'MAE': [4.575, 2.345, 2.129, 4.414, 4.353, 2.281, 2.511],
    'Runtime [ms]': ['0:00:40.055854', '0:11:55.468473', '0:01:03.575033', '0:00:29.231719',
                     '0:02:02.258870', '0:07:16.526615', '0:13:25.084937']
}

# Convert data to DataFrames
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df3 = pd.DataFrame(data3)

# Function to process DataFrame: calculate runtime in seconds and ranks
def process_df(df):
    df['Runtime [s]'] = pd.to_timedelta(df['Runtime [ms]']).dt.total_seconds()
    df['Rank'] = df['MAE'].rank()  # Rank models based on MAE (lower is better)
    return df

df1 = process_df(df1)
df2 = process_df(df2)
df3 = process_df(df3)

# Combine all data
combined = pd.concat([df1, df2, df3])

# Compute average rank and runtime for each model
agg_df = combined.groupby('Models', as_index=False).agg({
    'Rank': 'mean',
    'Runtime [s]': 'mean'
})

# Calculate frequencies for top N models in each DataFrame
top_n = 3
top_models = pd.concat([
    df1.nsmallest(top_n, 'MAE')['Models'],
    df2.nsmallest(top_n, 'MAE')['Models'],
    df3.nsmallest(top_n, 'MAE')['Models']
])
```

```python
# Calculate absolute frequencies for top 3 models
absolute_frequency_top3 = top_models.value_counts()

# Calculate absolute frequency for the top 1 model
top_model = pd.concat([
    df1.nsmallest(1, 'MAE')['Models'],
    df2.nsmallest(1, 'MAE')['Models'],
    df3.nsmallest(1, 'MAE')['Models']
])
absolute_frequency_top1 = top_model.value_counts()

# Create a combined DataFrame for plotting
all_models = absolute_frequency_top3.index.union(absolute_frequency_top1.index)
plot_df = pd.DataFrame(index=all_models)
plot_df['Top 3 Frequency'] = absolute_frequency_top3
plot_df['Top 1 Frequency'] = absolute_frequency_top1
plot_df = plot_df.fillna(0)


plot_df
```
